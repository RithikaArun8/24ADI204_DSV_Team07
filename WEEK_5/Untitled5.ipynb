```python
import pandas as pd
import numpy as np

filename = next(iter(uploaded))

df = pd.read_csv(filename, low_memory=False, encoding="latin1")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
display(df.head())
```

```python
# Dataset size
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Column names and data types
df.info()

# Missing values in each column
print(df.isnull().sum().sort_values(ascending=False).head(15))

# Summary of numerical columns
display(df.describe())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
# Remove columns that are completely empty
df = df.dropna(axis=1, how="all")

# Remove Order Zipcode because most values are missing
if "Order Zipcode" in df.columns:
    df = df.drop(columns=["Order Zipcode"])

    print("Columns remaining:", df.shape[1])
```

```python
# Fill missing text values with Unknown
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna("Unknown")

    # Fill missing numerical values with the median
    for col in df.select_dtypes(include=np.number).columns:
        if df[col].isnull().any():
                df[col] = df[col].fillna(df[col].median())

                print("Total missing values:", df.isnull().sum().sum())
```

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Remove extra spaces from text values
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()

    print("Text cleaning completed!")
```

```python
date_columns = [
      "order date (DateOrders)",
          "shipping date (DateOrders)"
          ]

for col in date_columns:
    if col in df.columns:
       df[col] = pd.to_datetime(df[col], errors="coerce")

print(df[date_columns].dtypes)
display(df[date_columns].head())
```

```python
private_columns = [
      "Customer Email",
      "Customer Password",
      "Customer Fname",
      "Customer Lname",
      "Customer Street"
]

df = df.drop(
columns=[col for col in private_columns if col in df.columns]
)

print("Personal information removed.")
print("Remaining columns:", df.shape[1])
```

```python
# Check data types
print(df.dtypes)

# Check numerical statistics
display(df.describe())

# Check category values
for col in ["Type", "Delivery Status", "Shipping Mode", "Order Status"]:
    if col in df.columns:
       print("\n", col)
       print(df[col].value_counts().head(10))
```

```python
numeric_columns = [
      "Sales",
      "Order Item Quantity",
      "Order Item Profit Ratio",
      "Days for shipping (real)"
]

for col in numeric_columns:
    if col in df.columns:
       q1 = df[col].quantile(0.25)
       q3 = df[col].quantile(0.75)
       iqr = q3 - q1

       lower = q1 - 1.5 * iqr
       upper = q3 + 1.5 * iqr

       count = ((df[col] < lower) | (df[col] > upper)).sum()

       print(f"{col}: {count} potential outliers")
```

```python
# Final validation
print("Final dataset shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

# Save as CSV
output_file = "DataCoSupplyChain_Cleaned.csv"
df.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
```

```python
# Handle postal codes separately
if "Customer Zipcode" in df.columns:
    df["Customer Zipcode"] = (
      df["Customer Zipcode"]
          .astype("string")
          .replace("nan", pd.NA)
          .fillna("Unknown")
)
# Fill missing text values
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna("Unknown")

# Fill missing numerical values, excluding postal codes
for col in df.select_dtypes(include=np.number).columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

print("Missing values:", df.isnull().sum().sum())
```
