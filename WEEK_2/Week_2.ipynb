importing libraries

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
```

loading dataset and printing first 5 rows


```python


df = pd.read_csv('/DataCoSupplyChainDataset.csv', encoding='latin1')
print("Dataset loaded successfully")
df.head()
```

Printing data types

```python
df.isnull().sum()
```

Checking duplicate values

```python

duplicates = df[df.duplicated()]

print("Number of duplicate rows:", len(duplicates))
print(duplicates)
```

Checking Statistics, categorical values and shape

```python

print("Dataset Shape:", df.shape)


print("\nNumerical Statistics:")
display(df.describe())


print("\nCategorical Values:")
display(df.describe(include=['object', 'category']))
```

```python
for col in df.select_dtypes(include=['object', 'category']).columns:
      counts = df[col].value_counts()

      if not counts.empty:
          print(f"\n{col}:")
          print(counts)
```

binary columns

```python

binary_columns = [
    col for col in df.columns
        if df[col].nunique(dropna=True) == 2
]
print("Binary Columns:", binary_columns)

display(df[binary_columns].head())
```

Initial Information

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

sns.histplot(df['Order Item Quantity'], bins=20, ax=axes[0])
axes[0].set_title('Order Quantity Distribution')

sns.histplot(df['Sales'], bins=20, ax=axes[1])
axes[1].set_title('Sales Distribution')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

df['Order Status'].value_counts().head(10).plot(
    kind='bar', ax=axes[0]
)
axes[0].set_title('Order Status Distribution')
axes[0].tick_params(axis='x', rotation=45)

df['Shipping Mode'].value_counts().plot(
    kind='bar', ax=axes[1]
)
axes[1].set_title('Shipping Mode Distribution')
axes[1].tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```
